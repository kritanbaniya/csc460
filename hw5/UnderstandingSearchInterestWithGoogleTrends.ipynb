The Data
The Google Trends data is available as a CSV file ai_tools_comparison.csv. The data contains the number of worldwide searches for chatGpt, Gemini, and Microsoft Copilot over the past 12 months as of September 2024.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data
trends = pd.read_csv('ai_tools_comparison.csv')

# Inspect the data
trends.head()
```

```python
# Start coding here
# Use as many cells as you need
```

```python
# Make 'week' a date and use it as the index
trends['week'] = pd.to_datetime(trends['week'])
trends.set_index('week', inplace=True)
```

```python
# Q1: Find the tool with the most consistent growth
# Weekly percentage change for each tool
growth_rates = trends.pct_change().fillna(0).mul(100)

# Lower standard deviation means more consistent growth
std_dev = growth_rates.std()

# Tool with the lowest standard deviation
most_consistent_tool = std_dev[std_dev == std_dev.min()].index[0]

print(f"Most consistent tool: {most_consistent_tool}")
```

```python
# Q2: Find when ChatGPT had its biggest drop
# Plot interest over time for each tool
trends.plot(title="Interest Levels of ChatGPT, Gemini, and Microsoft Copilot Over Time")
plt.xlabel('Week')
plt.ylabel('Interest Levels')
plt.show()

# Lowest point on the ChatGPT line
gpt_dip = "December 2023"

print(f"ChatGPT's biggest drop: {gpt_dip}")
```

```python
# Q3: Find the month with the highest average interest
# Convert weekly data to monthly averages
monthly_data = trends.resample('M').mean()

# Average across all tools, then pick the top month
best_month = monthly_data.mean(axis=1).idxmax().strftime('%B')

print(f"Month with the highest interest: {best_month}")
```
